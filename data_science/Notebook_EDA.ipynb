## 1. Configuración del entorno


```python
# Instalación de dependencias (Colab ya trae pandas/numpy/matplotlib/seaborn)
!pip install -q kaggle faker
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import subprocess

sns.set_theme(style="whitegrid")
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

RAW_DIR = Path("data/raw")
PROCESSED_DIR = Path("data/processed")
RAW_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

CATEGORIAS_PROYECTO = [
    "Alimentación", "Transporte", "Salud", "Vivienda",
    "Educación", "Ocio", "Servicios", "Otros",
]

print("Entorno configurado.")
```

## 2. Construir el dataset híbrido (llama al pipeline refactorizado)


```python
# Esta celda invoca el pipeline refactorizado `src.data.build_dataset_financeai`
# Si prefieres, puedes ejecutar el script localmente en tu máquina en lugar de desde el notebook.

RUTA_CAPA1 = RAW_DIR / "financial_transactions.csv"
RUTA_CAPA2 = RAW_DIR / "personal_finance_ml.csv"

if RUTA_CAPA1.exists() and RUTA_CAPA2.exists():
    print("Archivos crudos encontrados: ejecutando pipeline...")
    cmd = [sys.executable, "-m", "src.data.build_dataset_financeai",
           "--raw-dir", str(RAW_DIR), "--processed-dir", str(PROCESSED_DIR),
           "--n-users", "500", "--seed", str(RANDOM_SEED)]
    result = subprocess.run(cmd)
    if result.returncode == 0:
        print("Pipeline ejecutado correctamente.")
    else:
        print("El pipeline devolvió código de error:", result.returncode)
else:
    print("No se encontraron los CSV crudos en data/raw/.")
    print("Sube los archivos (financial_transactions.csv y personal_finance_ml.csv) a data/raw/ o ejecuta el script localmente.")
```

## 3. Cargar dataset final


```python
ruta_final = PROCESSED_DIR / "financeai_dataset_hibrido.csv"
if ruta_final.exists():
    df = pd.read_csv(ruta_final)
    print(f"Dataset cargado: {df.shape[0]} usuarios, {df.shape[1]} columnas")
else:
    print("No se encontró el dataset procesado. Ejecuta el pipeline o sube el CSV a data/processed/.")

# Muestra una vista previa si existe
try:
    df.head()
except NameError:
    pass
```

## 5. Visualizaciones (guardadas en outputs/figures/)

Las figuras se guardan en outputs/figures/ para referencia y uso en informes.


```python
# Visualizaciones: 1) Distribución ingreso_mensual, 2) Boxplot gasto_total por perfil,
# 3) Heatmap de correlaciones numéricas, 4) Top categorías de gasto agregadas.
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

fig_dir = Path('outputs/figures')
fig_dir.mkdir(parents=True, exist_ok=True)

if 'df' not in globals():
    if ruta_final.exists():
        df = pd.read_csv(ruta_final)
    else:
        print('Dataset procesado no encontrado. Ejecuta el pipeline primero.')

# 1) Distribución de ingreso_mensual (hist + box)
if 'ingreso_mensual' in df.columns:
    plt.figure(figsize=(10,6))
    plt.subplot(2,1,1)
    sns.histplot(df['ingreso_mensual'].dropna(), bins=30, kde=True, color='C0')
    plt.title('Distribución de ingreso_mensual')
    plt.subplot(2,1,2)
    sns.boxplot(x=df['ingreso_mensual'].dropna(), color='C1')
    plt.title('Boxplot de ingreso_mensual')
    plt.tight_layout()
    plt.savefig(fig_dir / 'ingreso_mensual_dist_box.png', dpi=150)
    plt.show()
else:
    print('Columna ingreso_mensual no encontrada en df')

# 2) Boxplot gasto_total por perfil_financiero
if 'gasto_total' in df.columns and 'perfil_financiero' in df.columns:
    df['gasto_total_log1p'] = np.log1p(df['gasto_total'])
    plt.figure(figsize=(8,5))
    sns.boxplot(x='perfil_financiero', y='gasto_total_log1p', data=df, palette='Set2')
    plt.ylabel('log1p(gasto_total)')
    plt.title('Gasto total (log1p) por perfil financiero')
    plt.tight_layout()
    plt.savefig(fig_dir / 'gasto_total_by_perfil.png', dpi=150)
    plt.show()
else:
    print('Columnas necesarias para boxplot no encontradas')

# 3) Heatmap de correlaciones numéricas
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if len(num_cols) >= 2:
    corr = df[num_cols].corr()
    plt.figure(figsize=(10,8))
    sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', square=False, linewidths=0.5)
    plt.title('Correlación entre variables numéricas')
    plt.tight_layout()
    plt.savefig(fig_dir / 'numeric_correlations_heatmap.png', dpi=150)
    plt.show()
else:
    print('No hay suficientes columnas numéricas para calcular correlación')

# 4) Top categorías de gasto agregadas (suma por columna de categorías)
cats = [c for c in CATEGORIAS_PROYECTO if c in df.columns]
if cats:
    sums = df[cats].sum().sort_values(ascending=False)
    plt.figure(figsize=(8,5))
    sns.barplot(x=sums.values, y=sums.index, palette='Blues_r')
    plt.xlabel('Gasto total (sum)')
    plt.title('Gasto total por categoría (agregado)')
    plt.tight_layout()
    plt.savefig(fig_dir / 'gasto_por_categoria.png', dpi=150)
    plt.show()
else:
    print('No se encontraron columnas de categorías en el dataset')
```
